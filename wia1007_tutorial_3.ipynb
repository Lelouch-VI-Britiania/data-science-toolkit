```python

import numpy as np
array_2x2 = np.array([[1, 2],
            [3, 4]])
print("1. Numpy array (2x2):")
print(array_2x2)
print("\n")
```

```python

random_3x4 = np.random.rand(3, 4)
print("2. Random numpy array (3x4):")
print(random_3x4)
print("\n")
```

```python
ones_array = np.ones((2, 3))
zeros_array = np.zeros((2, 3))
print("3. Array of Ones (2x3):")
print(ones_array)
print("Array of Zeroes (2x3):")
print(zeros_array)
print("\n")
```

```python
random_6x3 = np.random.rand(6, 3)
transposed_array = random_6x3.T
print("4. Random array (6x3):")
print(random_6x3)
print("Transposed array (3x6):")
print(transposed_array)
print("\n")
```

```python
n, p, m = 2, 3, 4  # example sizes
A = np.random.rand(n, p)
B = np.random.rand(p, m)
dot_product = np.dot(A, B)

print("5. Matrix A (n x p):")
print(A)
print("Matrix B (p x m):")
print(B)
print("Dot product (A · B) -> (n x m):")
print(dot_product)
```
